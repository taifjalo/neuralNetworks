# Week 2 Assignment: Graphical Monitoring and Overfitting

## California Housing Regression

This notebook addresses the assignment objectives:

1. Compare fully connected neural-network architectures.
2. Deliberately identify a sufficiently large unregularized model that overfits.
3. Compare regularization strategies using validation MAE.
4. Retrain the selected model using the training and validation sets combined.
5. Evaluate the final model once on the independent test set.

The test set is kept completely separate from model selection.

## 1. Loading the large dataset

The assignment specifically requires the `large` California Housing dataset. It contains 20,640 samples, eight numerical features, and one continuous target: the median house value. Twenty percent of the data is reserved as an independent test set.

```python
import numpy as np
import matplotlib.pyplot as plt

from keras.datasets import california_housing
from keras.models import Sequential
from keras.layers import Input, Dense, Dropout
from keras.regularizers import l2
from keras.callbacks import EarlyStopping
from keras.losses import MeanAbsoluteError
from keras.utils import set_random_seed

# Make the experiments as reproducible as possible.
set_random_seed(42)

(inputs, targets), (x_test, y_test) = california_housing.load_data(version='large', test_split=0.20, seed=42)

print('Input data:', inputs.shape, '/ Input targets:', targets.shape)
print('Test data:', x_test.shape, '/ Test targets:', y_test.shape)
```

## 2. Training, validation, and test sets

The test set is supplied by Keras and is not used during model selection. The remaining 16,512 samples are shuffled with a fixed seed and divided into 13,210 training samples and 3,302 validation samples.

The validation set is used to compare architectures and regularization methods. The test set is used only for the final performance estimate.

```python
# Shuffle only the non-test data before creating training and validation sets.
rng = np.random.default_rng(42)
indices = rng.permutation(len(inputs))
inputs = inputs[indices]
targets = targets[indices]

n_train = int(0.80 * len(inputs))
x_train = inputs[:n_train]      # [:13,210]
y_train = targets[:n_train]     # [:13,210]
x_val = inputs[n_train:]        # [13,210:]
y_val = targets[n_train:]       # [13,210:]

print('Training:', x_train.shape, y_train.shape)
print('Validation:', x_val.shape, y_val.shape)

print('Test:', x_test.shape, y_test.shape)
```

## 3. Preprocessing without data leakage

Each input feature is standardized using the mean and standard deviation computed from the training set only:

$$z = (x - \mu_{train}) / \sigma_{train}$$

The same training-set statistics are then applied to validation and test data. This prevents information from the validation or test sets leaking into training. The target is divided by 100,000 to keep its numerical scale close to one.

```python
mean = x_train.mean(axis=0)
std = x_train.std(axis=0)

x_train = (x_train - mean) / std
x_val = (x_val - mean) / std
x_test = (x_test - mean) / std

y_train = y_train / 1e5
y_val = y_val / 1e5
y_test = y_test / 1e5

print('Training feature means after standardization:', np.round(x_train.mean(axis=0), 3))
print('Training feature stds after standardization:', np.round(x_train.std(axis=0), 3))
```

## 4. Finding an overfitting architecture

The first experiments use no regularization. A small model is compared with a much larger model. The purpose of the large model is to have enough capacity to fit the training set very closely, so that overfitting can be recognized from the training and validation curves.

For this regression problem, the output layer has one linear unit. The loss is MSE and MAE is monitored because it is easier to interpret in the original target units.

```python
def compile_regression_model(model):
    model.compile(optimizer='adam', loss='mse', metrics=['mae'])
    return model

small_model = compile_regression_model(Sequential([
    Input(shape=(8,)),
    Dense(32, activation='relu'),
    Dense(1)
]))

large_model = compile_regression_model(Sequential([
    Input(shape=(8,)),
    Dense(128, activation='relu'),
    Dense(128, activation='relu'),
    Dense(128, activation='relu'),
    Dense(1)
]))

# No regularization is used in this stage to train the models.
history_small = small_model.fit(x_train, y_train, epochs=100, batch_size=32, validation_data=(x_val, y_val), verbose=0)

history_large = large_model.fit(x_train, y_train, epochs=100, batch_size=32, validation_data=(x_val, y_val), verbose=0)


# Print the final training MAE and the best validation MAE for both models
for name, history in [('Small', history_small), ('Large', history_large)]:
    val_mae = np.array(history.history['val_mae'])
    train_mae = np.array(history.history['mae'])
    print(f'{name} final training MAE: {train_mae[-1]:.4f}')
    print(f'{name} best validation MAE: {val_mae.min():.4f} at epoch {val_mae.argmin() + 1}')
```

```python
# Plot the training and validation loss for both small and large models
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history_small.history['mae'], label='Small training MAE')
axes[0].plot(history_small.history['val_mae'], label='Small validation MAE')
axes[0].set_title('Small unregularized model')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('MAE (scaled target)')
axes[0].legend()

axes[1].plot(history_large.history['mae'], label='Large training MAE')
axes[1].plot(history_large.history['val_mae'], label='Large validation MAE')
axes[1].set_title('Large unregularized model')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('MAE (scaled target)')
axes[1].legend()

plt.tight_layout()
plt.show()
```

### Interpretation of the overfitting experiment

The large unregularized model is considered overfitted when its training MAE continues to decrease while its validation MAE stops improving and starts to increase. The training-validation gap shows that the model is fitting details specific to the training samples rather than learning relationships that generalize to unseen samples.

This is why the final training MAE is not used to select the model. The relevant quantity for model selection is the best validation MAE.

## 5. Comparing regularization strategies

Two common strategies are tested:

- **L2 regularization:** adds a penalty for large weights.
- **Dropout:** randomly disables a fraction of hidden-unit outputs during training.

The two models are compared using their minimum validation MAE. The test set is not used in this comparison. Early stopping limits unnecessary computation while keeping the best weights found during each experiment.

```python
def make_l2_model():
    return compile_regression_model(Sequential([
        Input(shape=(8,)),
        Dense(64, activation='relu', kernel_regularizer=l2(0.001)),
        Dense(64, activation='relu', kernel_regularizer=l2(0.001)),
        Dense(1)
    ]))

def make_dropout_model():
    return compile_regression_model(Sequential([
        Input(shape=(8,)),
        Dense(64, activation='relu'),
        Dropout(0.20),
        Dense(64, activation='relu'),
        Dropout(0.20),
        Dense(1)
    ]))

def train_regularized(model):
    callback = EarlyStopping(
        monitor='val_mae', patience=10, mode='min',
        restore_best_weights=True
    )
    return model.fit(x_train, y_train, epochs=150, batch_size=32, validation_data=(x_val, y_val), callbacks=[callback], verbose=0)

l2_model = make_l2_model()
history_l2 = train_regularized(l2_model)

dropout_model = make_dropout_model()
history_dropout = train_regularized(dropout_model)

l2_val_mae = np.array(history_l2.history['val_mae'])
dropout_val_mae = np.array(history_dropout.history['val_mae'])

best_l2 = l2_val_mae.min()
best_dropout = dropout_val_mae.min()
epoch_l2 = l2_val_mae.argmin() + 1
epoch_dropout = dropout_val_mae.argmin() + 1

print(f'L2 best validation MAE: {best_l2:.4f} at epoch {epoch_l2}')
print(f'Dropout best validation MAE: {best_dropout:.4f} at epoch {epoch_dropout}')
```

```python
plt.figure(figsize=(10, 5))
plt.plot(l2_val_mae, label='L2 validation MAE')
plt.plot(dropout_val_mae, label='Dropout validation MAE')
plt.xlabel('Epoch')
plt.ylabel('Validation MAE (scaled target)')
plt.title('Regularization comparison')
plt.legend()
plt.show()
```

```python
if best_l2 <= best_dropout:
    selected_model_name = 'L2'
    selected_epochs = epoch_l2
    selected_validation_mae = best_l2
else:
    selected_model_name = 'Dropout'
    selected_epochs = epoch_dropout
    selected_validation_mae = best_dropout

print(f'Selected model: {selected_model_name}')
print(f'Selected validation MAE: {selected_validation_mae:.4f}')
print(f'Selected number of epochs: {selected_epochs}')
```

## 6. Final training and independent testing

After selecting the architecture and regularization method using only the training and validation data, those two sets are combined. A fresh instance of the selected model is then trained once more. The test set remains untouched until the final evaluation.

The final MAE is reported both in scaled units and in dollars. Since the target was divided by 100,000, the conversion is:

$$MAE_{dollars} = MAE_{scaled} \times 100000$$

```python
x_train_full = np.concatenate((x_train, x_val), axis=0)
y_train_full = np.concatenate((y_train, y_val), axis=0)

if selected_model_name == 'L2':
    final_model = make_l2_model()
else:
    final_model = make_dropout_model()

final_model.fit(
    x_train_full, y_train_full,
    epochs=selected_epochs, batch_size=32, verbose=0
)

test_loss, test_mae = final_model.evaluate(x_test, y_test, verbose=0)
print(f'Final test MAE: {test_mae:.4f} in scaled units')
print(f'Final test MAE: ${test_mae * 1e5:,.0f}')
```

## 7. Comparison with a trivial baseline

As a sanity check, the neural network is compared with a trivial predictor that always predicts the mean target value of the combined training data. A useful model should have a lower test MAE than this baseline.

```python
baseline_prediction = np.full_like(y_test, np.mean(y_train_full))
mae_function = MeanAbsoluteError()
baseline_mae = float(mae_function(y_test, baseline_prediction).numpy())
improvement = 100 * (baseline_mae - test_mae) / baseline_mae

print(f'Baseline test MAE: {baseline_mae:.4f} in scaled units')
print(f'Baseline test MAE: ${baseline_mae * 1e5:,.0f}')
print(f'Improvement over baseline: {improvement:.2f}%')
```

## Conclusion

The large unregularized network was used to demonstrate overfitting by comparing training and validation curves. L2 regularization and dropout were then compared using the minimum validation MAE, without using the test set. The better regularized model was selected, retrained from scratch on the combined training and validation data, and evaluated once on the independent test set. The baseline comparison confirms whether the final neural network improves substantially over a simple constant prediction.
